## Importação das Bibliotecas

```python
import requests
import pandas as pd
import json
```

## COLETA DA API REST

```python
#criação de uma variável para armazenar a url que contém os dados da API
#Uso da biblioteca requests com o método GET para obter os dados da API

url = r'http://127.0.0.1:5000/country'
r = requests.get(url)
print(r.status_code)                            #exibe o status da requisição feita para a API para verificar se foi realizada com sucesso ('CÓDIGO: 200' indica que deu certo)
print(r.json())                                 #exibe o conteúdo da API em formato json
```

```python
#Criação de um DataFrame com os dados obtidos da API
#Uso do 'orient=index' devido ao json possuir a estrutura com índices
#transformação do índice, que se refere aos código dos países, em uma coluna no DataFrame com a nomenclatura 'code'


df_api_original = pd.DataFrame.from_dict(dict(r.json()), orient='index').reset_index().rename(columns={'index':'code'})
df_api = df_api_original.copy()   #criação de uma cópia do DataFrame para manipulação dos dados e presevação do original
```

```python
#exibição do DataFrame para melhor visualização dos dados obtidos com a API

df_api
```

```python
#Verificação do tipo de dado de cada coluna e da quantidade de valores não nulos

df_api.info()
```

## COLETA DO ARQUIVO CSV

```python
#Leitura do arquivo CSV que contém os dados do PIB e criação de um DataFrame com os dados

df_arq_original = pd.read_csv('gdp_data.csv')
df_arq = df_arq_original.copy()                         #criação de uma cópia do DataFrame para manipulação dos dados e presevação do original
```

```python
#exibição do DataFrame para melhor visualização dos dados

df_arq
```

```python
#Verificação do tipo de dado de cada coluna e da quantidade de valores não nulos

df_arq.info()
```

```python
#remoção das colunas que não serão utilizadas.

colunas_remover = ['Indicator Name','Indicator Code','Unnamed: 66']
df_arq = df_arq.drop(columns=colunas_remover)
df_arq                              #exibição do DataFrame após a remoção das coluna para verificar se ficou correto
```

## CONSOLIDAÇÃO DOS DADOS (API REST + CSV)

```python
#junção do DataFrame com os dados da API REST e do DataFrame com os dados do CSV em um único DataFrame, com uso da função 'merge'
#uso do parâmetro 'how=left' que pega todos os valores do DataFrame com os dados da API e junta com o DataFrame com os dados do CSV.
#junção realizada a partir da coluna 'code' que pertence ao df_api e da coluna 'Country Code' que pertencem ao df_arq, que representam o código do país (coluna em comum entre os DataFrames)
#registros que constam no CSV, mas não constam na API são desconsiderados 



df_unificado = pd.merge(df_api,df_arq, how = 'left', left_on='code', right_on='Country Code')
df_unificado                             #exibição do DataFrame após a junção para verificar se ficou correto
```

```python
#Verificação da quantidade de valores não nulos
df_unificado.info()
```

```python
#verificação das colunas que possuem valores ausentes

df_unificado.isna()
```

```python
#verificação da quantidade de valores ausentes na coluna 'Country Name' que pertence ao dados extraídos do CSV
#a partir dessa consulta foi possível confirmar que há registros que constam apenas na API, visto que foram identificados valores ausentes na coluna 'Country Name' que pertence aos dados coletados do CSV

df_unificado['Country Name'].isna().sum()
```

```python
#verificação da quantidade de valores ausentes na coluna 'Country Code' que pertence ao dados extraídos do CSV
#a partir dessa consulta foi possível confirmar que há registros que constam apenas na API, visto que foram identificados valores ausentes na coluna 'Country Code' que pertence aos dados coletados do CSV

df_unificado['Country Code'].isna().sum()
```

```python
#remoção das linhas que possuem valores ausentes nas colunas 'Country Name' e 'Country Code'

df_unificado = df_unificado.dropna(subset=['Country Name','Country Code'])
df_unificado
```

```python
#remoção da coluna 'Country Code' e 'Country Name', pois as colunas 'code' e 'name' possuem as mesmas informações

colunas_remover_unificado = ['Country Code','Country Name']
df_unificado = df_unificado.drop(columns=colunas_remover_unificado)
df_unificado
```

```python
#Verificação da quantidade de valores não nulos após a unificação e ajustes do DataFrame

df_unificado.info()
```

```python
#Criação de um arquivo CSV gerado a partir do DataFrame unificado que possui todas as informações dos países

df_unificado.to_csv('base_unificada.csv', index=False, encoding='utf-8')
```
