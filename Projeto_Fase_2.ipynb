# **5. Consolidação das bases de dados em um DataFrame**

Após a análise dos conjuntos de dados e a padronização das colunas, realizou-se a consolidação das bases em um único DataFrame. Para isso, utilizou-se a função pd.concat para empilhar as bases verticalmente, uma abaixo da outra, já que todas possuem as mesmas colunas.

```python
#Criação de um DataFrame unificando todas as bases de 2016 a 2020

df_OSMI_unificado_original = pd.concat(
    [df_2016, df_2017, df_2018, df_2019, df_2020],
    axis=0,                                #realiza a concatenação dos Dataframes em um único (empilha os dados verticalmente)
    ignore_index=True)                   #ignora o índices das bases e reinicia a partir de 0

df_OSMI_unificado = df_OSMI_unificado_original.copy()       #criação de uma cópia para manipulação dos dados e preservação do original

df_OSMI_unificado                          #exibe os DataFrame unificado
```

Como esta variável é necessária para responder as questões do estudo de caso, optou-se por padronizar ambas como "Não elegível – NA" para que os dados ficassem consistentes.

```python
#criação de um boxplot para verificação possíveis outliers

df_OSMI_unificado['P11'].plot.box(figsize=(12,4))
```

```python
#cálculo do IQR
Q1 = df_OSMI_unificado['P11'].quantile(0.25)
Q3 = df_OSMI_unificado['P11'].quantile(0.75)
IQR = Q3 - Q1

#cria uma máscara com valores de idade considerados outliers pelo IQR
mask_outliers = (df_OSMI_unificado['P11'] > (Q3+1.5 *IQR)) | (df_OSMI_unificado['P11'] < (Q1-1.5 *IQR))

#cria uma lista com os valores de idade identificados como possíveis outliers (a partir da máscara)
outliers_idade = df_OSMI_unificado.loc[mask_outliers, 'P11'].unique()
outliers_idade      #exibe a lista com os possíveis outliers
```

```python
#definição dos valores outliers do campo idade
#serão retirados apenas os valores (0, 3, 5, 15, 17 99 e 323)
#as demais idades apresentadas pelo boxplot serão mantidas, pois são idades aceitáveis e comuns para pessoas que trabalham

#cria uma máscara para os valores fora do intervalo de 18 a 74 anos
mask_outliers_idade = (df_OSMI_unificado['P11'] < 18) | (df_OSMI_unificado['P11'] > 74)
```

```python
#criar uma cópia do DataFrame mantendo apenas as linhas com idades dentro do intervalo válido

df_OSMI_unificado = df_OSMI_unificado[~mask_outliers_idade ].copy()
```

```python
#verificação dos valores únicos da coluna 'P11' após a retirada dos outliers e alteração do tipo de dado para conferência

df_OSMI_unificado['P11'].unique()
```

```python
#PEGANDO OS DADOS DE DOENÇA MENTAL, ANO E PAÍS

#cria uma máscara para identificar os casos com diagnóstico de doença mental igual a 'Sim' e ano de aplicação da pesquisa '2016'
mask_doenca_mental_2016 = (df_unificado_pib['P15'] == 'Sim') & (df_unificado_pib['P17'] == 2016)

#Cria um DataFrame a partir da máscara 'mask_doenca_mental_2016' a um DataFrame, ou seja, apenas com os casos que atendem a condição da máscara
df_doenca_mental_2016 = df_unificado_pib[mask_doenca_mental_2016]

#cálcula o percentual de pessoas com doença mental em cada país no ano de 2016 a partir do DataFrame criado
percentuais_dm_2016 = ((df_doenca_mental_2016['P14']
                       .value_counts(normalize=True) * 100)
                       .round(1))

#PEGANDO OS DADOS DO PAÍS E PIB DE 2016

#crio uma máscara dos registros que o ano de aplicação da pesquisa é igual a 2016
mask_2016 = (df_unificado_pib['P17'] == 2016)

#atribuo a máscara 'mask_pib_pais_2016' em DataFrame apenas com as colunas de país e pib, apenas com os casos que atendem a condição da máscara
#uso do drop_duplicates para desconsiderar os duplicados, visto que o dado do país e pib se repetem
pib_pais_2016 = (df_unificado_pib[mask_2016][['P14', '2016']]
                 .drop_duplicates())


#UNIFICANDO OS DADOS

print('Comparativo - (%) Diagnósticos de doença mental por País (ano 2016)')   #exibe o título para a tabela

#converte a série percentuais_dm_2016 em um DataFrame e unifica com o dado de pib de cada país
df_pib_doenca = (percentuais_dm_2016
                 .to_frame()
                 .merge(pib_pais_2016, left_on='P14', right_on='P14', how='left'))

#cria uma variável para renomear as colunas
colunas_renomeadas = (df_pib_doenca
                      .rename(columns= {'P14':'País',
                                        'proportion':'Doença Mental (%)',
                                        '2016': 'PIB (2016)'}))

#ordena as colunas
df_pib_doenca = (colunas_renomeadas[['País',
                                     'PIB (2016)',
                                     'Doença Mental (%)']])

#ordena o PIB do menor para o maior e reinicia o índice
(df_pib_doenca
 .sort_values('PIB (2016)', ascending=False)
 .reset_index(drop=True))

```
